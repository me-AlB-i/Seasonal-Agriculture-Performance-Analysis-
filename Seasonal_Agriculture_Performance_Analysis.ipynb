## 2.4 Handling Missing Values

The missing values identified in the dataset are handled using the median of the respective numerical columns. This helps retain the available observations without removing complete records.

```python
df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())
```

# 14. Key Findings

The analysis of seasonal agricultural performance revealed several important patterns:

1. Kharif recorded the highest average yield, production, revenue and profit among the three seasons.

2. Zaid recorded negative average profit despite generating considerable revenue, indicating that higher revenue does not necessarily result in higher profitability.

3. Kharif experienced the highest average rainfall, humidity and soil moisture, while Zaid had the highest average temperature and sunlight hours.

4. Environmental factors showed very weak linear correlations with crop yield in the dataset.

5. Water usage showed the strongest correlation with crop yield among the analyzed agricultural factors, with a correlation of approximately 0.39.

6. Drip irrigation recorded the highest average crop yield, while Rainfed agriculture recorded the highest average water efficiency.

7. Sugarcane showed the highest average yield and average profit among the crops analyzed.

8. Punjab recorded the highest average crop yield and was also among the highest-performing states in terms of average profit.

9. Soil pH, fertilizer usage and pesticide usage showed relatively small differences across seasons.

10. Kharif recorded the highest average disease and pest risk among the three seasons.

# 15. Recommendations

Based on the findings from the dataset, the following recommendations can be considered:

- Seasonal agricultural planning should give particular attention to Kharif, which showed strong overall performance but also higher disease and pest risk.

- Efficient irrigation methods should be considered to improve water utilization while maintaining crop productivity.

- Crop selection can be guided by both yield and profitability rather than yield alone.

- Water usage should be monitored carefully because it showed a moderate relationship with crop yield.

- Farmers should consider regional differences when planning crop production and resource allocation.

- Disease and pest management should receive greater attention during seasons with higher observed risk.

- Further analysis using larger and more detailed datasets could help identify more specific relationships between environmental conditions, farming practices and crop performance.
